### Solution de l'exercice 1 · Signe d'un nombre

Le cas nul est traité par le else final : inutile d'écrire `x == 0` explicitement. Couvrir le cas par défaut avec else évite de laisser passer une valeur imprévue.

```python
def signe(x):
    """Retourne le signe de x sous forme de chaîne."""
    if x > 0:
        return "positif"
    elif x < 0:
        return "négatif"
    else:
        return "nul"

print(signe(5), signe(-3.5), signe(0))
```

```python
assert signe(5) == "positif"
assert signe(-3.5) == "négatif"
assert signe(0) == "nul"
print("Solution 1 validée par les tests.")
```

### Solution de l'exercice 2 · FizzBuzz sur une plage

Le cas le plus spécifique (multiple de 15) doit être testé en premier, sinon 15 serait capté par le test multiple de 3. C'est la même règle d'ordre que pour les mentions de la séance 01.

```python
def fizzbuzz(n):
    """Version classique de FizzBuzz de 1 à n."""
    resultat = []
    for k in range(1, n + 1):
        if k % 15 == 0:
            resultat.append("FizzBuzz")
        elif k % 3 == 0:
            resultat.append("Fizz")
        elif k % 5 == 0:
            resultat.append("Buzz")
        else:
            resultat.append(k)
    return resultat

print(fizzbuzz(15))
```

```python
reponse = fizzbuzz(15)
assert reponse[2] == "Fizz"
assert reponse[4] == "Buzz"
assert reponse[14] == "FizzBuzz"
assert len(reponse) == 15
print("Solution 2 validée par les tests.")
```

### Solution de l'exercice 3 · Somme des nombres pairs avec while

En partant de 2 et en avançant de 2 en 2, la condition `k <= n` ne teste jamais un impair : plus efficace que de tester la parité à chaque tour. La mise à jour `k += 2` garantit la terminaison.

```python
def somme_pairs(n):
    """Somme des pairs de 1 à n inclus, en while obligatoire."""
    total = 0
    k = 2
    while k <= n:
        total += k
        k += 2
    return total

print(somme_pairs(10), somme_pairs(7))
```

```python
assert somme_pairs(10) == 30
assert somme_pairs(7) == 12
assert somme_pairs(1) == 0
assert somme_pairs(0) == 0
print("Solution 3 validée par les tests.")
```

### Solution de l'exercice 4 · Devinette avec compteur d'essais

Le `return` remplace ici le couple témoin et break : dès que le secret est trouvé, la fonction rend sa réponse et s'arrête. Si la boucle se termine sans return, c'est l'échec qui est renvoyé après la boucle.

```python
def devine(secret, essais):
    """Retourne le numéro du tour gagnant, ou -1 si échec."""
    tour = 0
    for essai in essais:
        tour += 1
        if essai == secret:
            return tour
    return -1

print(devine(42, [10, 50, 42, 7]), devine(99, [1, 2, 3]))
```

```python
assert devine(42, [10, 50, 42, 7]) == 3
assert devine(1, [1, 2, 3]) == 1
assert devine(99, [1, 2, 3]) == -1
assert devine(5, []) == -1
print("Solution 4 validée par les tests.")
```

### Solution de l'exercice 5 · Pyramide d'étoiles

Deux multiplications de chaînes remplissent le rôle des boucles imbriquées : `" " * (n - k)` pour la marge, `"*" * (2 * k - 1)` pour le corps. Vérifiez la formule sur k = 1 : une seule étoile, c'est le sommet.

```python
def pyramide(n):
    """Retourne les n lignes d'une pyramide d'étoiles centrée."""
    lignes = []
    for k in range(1, n + 1):
        espaces = " " * (n - k)
        etoiles = "*" * (2 * k - 1)
        lignes.append(espaces + etoiles)
    return lignes

for ligne in pyramide(5):
    print(ligne)
```

```python
assert pyramide(3) == ["  *", " ***", "*****"]
assert pyramide(1) == ["*"]
assert pyramide(0) == []
print("Solution 5 validée par les tests.")
```
