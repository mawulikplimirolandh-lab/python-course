### Solution de l'exercice 1 · Division sécurisée

Le try/except ne couvre que le risque annoncé. Si a est une chaîne, la division lève TypeError dans le try : aucune clause ne correspond, l'exception remonte naturellement à l'appelant. C'est le comportement demandé : n'attrapez que ce que vous savez traiter.

```python
def division_secu(a, b):
    """Retourne a / b, ou None si division par zéro."""
    try:
        return a / b
    except ZeroDivisionError:
        return None

print(division_secu(10, 2), division_secu(10, 0))
```

```python
assert division_secu(10, 2) == 5.0
assert division_secu(10, 0) is None
print("Solution 1 validée par les tests.")
```

### Solution de l'exercice 2 · Conversion robuste avec valeur de secours

Un seul except peut annoncer un tuple de types : `(ValueError, TypeError)`. int(None) lève TypeError, int("douze") lève ValueError : les deux incidents tombent dans le même filet et donnent le même secours.

```python
def en_entier(valeur, defaut=0):
    """Convertit en int, ou retourne la valeur de secours."""
    try:
        return int(valeur)
    except (ValueError, TypeError):
        return defaut

print(en_entier("42"), en_entier("douze", -1), en_entier(None))
```

```python
assert en_entier(42) == 42
assert en_entier("42") == 42
assert en_entier("douze", -1) == -1
assert en_entier(None, 3) == 3
print("Solution 2 validée par les tests.")
```

### Solution de l'exercice 3 · Compte bancaire avec exception métier

Deux règles, deux exceptions distinctes : ValueError pour une demande absurde, SoldeInsuffisantError pour une demande légitime mais impossible. Le caissier n'attrape QUE l'exception métier : une ValueError (montrant un bug d'appel) remonterait au grand jour au lieu d'être étouffée.

```python
class SoldeInsuffisantError(Exception):
    """Levée quand un retrait dépasse le solde disponible."""


class Compte:
    """Compte dont le retrait respecte deux règles métier."""

    def __init__(self, titulaire, solde):
        self.titulaire = titulaire
        self.solde = solde

    def retirer(self, montant):
        if montant <= 0:
            raise ValueError("le retrait doit être strictement positif")
        if montant > self.solde:
            raise SoldeInsuffisantError(
                f"retrait de {montant} F pour un solde de {self.solde} F")
        self.solde -= montant

def caissier(compte, demandes):
    """Tente chaque retrait, collecte les refus."""
    refus = []
    for montant in demandes:
        try:
            compte.retirer(montant)
        except SoldeInsuffisantError as e:
            refus.append(str(e))
    return refus

c = Compte("Awa", 1000)
print(caissier(c, [300, 900, 100]), c.solde)
```

```python
c = Compte("Awa", 1000)
refus = caissier(c, [300, 900, 100])
assert c.solde == 600
assert len(refus) == 1
try:
    c.retirer(0); refuse = False
except ValueError:
    refuse = True
assert refuse
print("Solution 3 validée par les tests.")
```

### Solution de l'exercice 4 · Lecture de fichier robuste avec finally

`ouvert = True` est placé APRÈS la lecture réussie : si open échoue, il n'est jamais atteint. Le finally montre l'emplacement garanti ; dans un code de production, on y journalise la tentative. with ferme le fichier même si read() plantait : deux niveaux de sûreté.

```python
def lire_fichier(nom):
    """Lit un fichier ; signale l'ouverture même en cas d'échec."""
    ouvert = False
    contenu = None
    try:
        with open(nom, "r") as f:
            contenu = f.read()
            ouvert = True
    except FileNotFoundError:
        contenu = "fichier absent"
    finally:
        pass   # le bloc garanti : ici, on pourrait journaliser
    return (contenu, ouvert)

print(lire_fichier("present.txt"))
print(lire_fichier("absent.txt"))

# Nettoyage du fichier de test
import os
os.remove("present.txt")
```

```python
with open("present.txt", "w") as f:
    f.write("ligne une\nligne deux")
contenu, ouvert = lire_fichier("present.txt")
assert ouvert is True and contenu == "ligne une\nligne deux"
contenu, ouvert = lire_fichier("absent.txt")
assert ouvert is False and contenu == "fichier absent"
import os
if os.path.exists("present.txt"):
    os.remove("present.txt")
print("Solution 4 validée par les tests.")
```

### Solution de l'exercice 5 · Validateur de données CSV

Le raise interne transforme chaque règle métier en ValueError unique, attrapée par le même except : le corps reste linéaire et lisible. `enumerate(..., start=1)` donne la numérotation humaine des erreurs. Une ligne fausse n'affecte jamais ses voisines : c'est le contrat du traitement de lot.

```python
def valider(lignes):
    """Sépare fiches valides et erreurs, ligne par ligne."""
    fiches = []
    erreurs = []
    for numero, ligne in enumerate(lignes, start=1):
        champs = ligne.split(";")
        try:
            if len(champs) != 3:
                raise ValueError("trois champs attendus")
            nom, age_texte, ville = champs
            age = int(age_texte)
            if age < 0:
                raise ValueError("age négatif")
            fiches.append({"nom": nom, "age": age, "ville": ville})
        except ValueError as e:
            erreurs.append(f"ligne {numero} : {e}")
    return (fiches, erreurs)

fiches, erreurs = valider(LIGNES)
print(len(fiches), "fiches,", len(erreurs), "erreurs")
for erreur in erreurs:
    print(" ", erreur)
```

```python
fiches, erreurs = valider(LIGNES)
assert len(fiches) == 2
assert fiches[0]["age"] == 25 and fiches[1]["ville"] == "Atakpamé"
assert len(erreurs) == 3
assert erreurs[0].startswith("ligne 2 :")
print("Solution 5 validée par les tests.")
```
