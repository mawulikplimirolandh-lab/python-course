### Solution de l'exercice 1 · Fiche contact

L'affectation `fiche["tel"] = ...` remplace une valeur existante, la même syntaxe crée « ville » : c'est le principe d'unification du dictionnaire. `sorted()` retourne une nouvelle liste triée, contrairement à `.sort()` qui trie sur place.

```python
def fiche_contact():
    """Construit une fiche, la met à jour et liste ses clés."""
    fiche = {"nom": "Awa", "tel": "90000001"}
    fiche["tel"] = "90111222"
    fiche["ville"] = "Lomé"
    return sorted(fiche)

print(fiche_contact())
```

```python
assert fiche_contact() == ["nom", "tel", "ville"]
print("Solution 1 validée par les tests.")
```

### Solution de l'exercice 2 · Compteur d'occurrences

Version condensée avec get : `compteur[mot] = compteur.get(mot, 0) + 1` remplace les quatre lignes du SI. Retenez les deux écritures : la longue pour comprendre, la courte pour produire.

```python
def compter(mots):
    """Compte les apparitions de chaque mot."""
    compteur = {}
    for mot in mots:
        if mot in compteur:
            compteur[mot] += 1
        else:
            compteur[mot] = 1
    return compteur

print(compter(["riz", "mais", "riz", "yam", "mais", "riz"]))
```

```python
assert compter(["riz", "mais", "riz", "yam", "mais", "riz"]) == {
    "riz": 3, "mais": 2, "yam": 1}
assert compter([]) == {}
print("Solution 2 validée par les tests.")
```

### Solution de l'exercice 3 · Validation d'adresse e-mail

`partition` découpe autour du premier @ et retourne trois morceaux : la partie locale, le séparateur et le domaine. La variable fantôme `_` reçoit le @ lui même, par convention elle signale qu'on n'utilise pas cette valeur. Une validation réelle passe par les expressions régulières, vues au Tome 2.

```python
def email_valide(adresse):
    """Validation simplifiée : @ unique, point après @, pas d'espace."""
    if len(adresse) < 6:
        return False
    if adresse.count("@") != 1:
        return False
    if " " in adresse:
        return False
    partie_locale, _, domaine = adresse.partition("@")
    return "." in domaine and len(domaine) > 1

print(email_valide("a.b@ecole.tg"))
```

```python
assert email_valide("a.b@ecole.tg") is True
assert email_valide("sans-arobase.tg") is False
assert email_valide("deux@@x.tg") is False
assert email_valide("x@sanspoint") is False
assert email_valide("a b@x.tg") is False
print("Solution 3 validée par les tests.")
```

### Solution de l'exercice 4 · Statistiques de notes en dictionnaire

Une compréhension collecte les notes admises, puis le dictionnaire se construit en un seul littéral lisible. `not notes` est la manière idiomatique de tester une liste vide. 3 admis sur 5 notes donnent un taux de 60.

```python
def statistiques(notes):
    """Résumé complet d'une série de notes en un dictionnaire."""
    if not notes:
        return {"erreur": "liste vide"}
    reussis = [n for n in notes if n >= 10]
    return {
        "min": min(notes),
        "max": max(notes),
        "moyenne": round(sum(notes) / len(notes), 2),
        "reussis": len(reussis),
        "taux": round(100 * len(reussis) / len(notes)),
    }

print(statistiques([12, 8, 15, 9, 14]))
```

```python
s = statistiques([12, 8, 15, 9, 14])
assert s == {"min": 8, "max": 15, "moyenne": 11.6,
             "reussis": 3, "taux": 60}
assert statistiques([]) == {"erreur": "liste vide"}
print("Solution 4 validée par les tests.")
```

### Solution de l'exercice 5 · Mini quiz en fonctions

`enumerate(QUIZ, start=1)` donne le numéro de question sans manipuler d'index. `reponses.get(numero)` rend None pour une question sans réponse : la comparaison échoue proprement, sans KeyError. Les deux fonctions partagent la même boucle : c'est normal et lisible.

```python
def corriger(reponses):
    """Score obtenu : nombre de réponses exactes."""
    score = 0
    for numero, donnee in enumerate(QUIZ, start=1):
        if reponses.get(numero) == donnee["bonne"]:
            score += 1
    return score

def detail(reponses):
    """Numéros des questions dont la réponse est fausse ou absente."""
    ratees = []
    for numero, donnee in enumerate(QUIZ, start=1):
        if reponses.get(numero) != donnee["bonne"]:
            ratees.append(numero)
    return ratees

reponses_awa = {1: 1, 2: 0, 3: 2}
print(corriger(reponses_awa), detail(reponses_awa))
```

```python
awa = {1: 1, 2: 0, 3: 2}
assert corriger(awa) == 1
assert detail(awa) == [2, 3]
assert corriger({1: 1, 2: 1, 3: 1}) == 3
assert detail({}) == [1, 2, 3]
print("Solution 5 validée par les tests.")
```
