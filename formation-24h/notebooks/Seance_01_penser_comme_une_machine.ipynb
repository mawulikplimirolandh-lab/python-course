### Solution de l'exercice 1 · Périmètre et aire d'un rectangle

On applique directement les deux formules. Le mot clé `return` renvoie les deux valeurs regroupées dans un tuple, que Python considère comme une seule réponse.

```python
def perimetre_aire(longueur, largeur):
    """Retourne (perimetre, aire) d'un rectangle."""
    perimetre = 2 * (longueur + largeur)
    aire = longueur * largeur
    return (perimetre, aire)

print(perimetre_aire(8, 5))
```

```python
assert perimetre_aire(8, 5) == (26, 40)
assert perimetre_aire(2.5, 4) == (13.0, 10.0)
assert perimetre_aire(1, 1) == (4, 1)
print("Solution 1 validée par les tests.")
```

### Solution de l'exercice 2 · Pair ou impair

L'expression `n % 2 == 0` se lit : le reste de n divisé par 2 est-il nul ? Sa valeur est déjà `True` ou `False`, donc un simple `return` suffit, sans SI.

```python
def est_pair(n):
    """Retourne True si n est pair, False sinon."""
    return n % 2 == 0

print(est_pair(4), est_pair(7))
```

```python
assert est_pair(4) is True
assert est_pair(7) is False
assert est_pair(0) is True
assert est_pair(-6) is True
print("Solution 2 validée par les tests.")
```

### Solution de l'exercice 3 · Trier trois nombres sans sorted()

La somme des trois valeurs moins le plus petit et le plus grand laisse exactement le milieu. Une version purement conditionnelle est aussi possible mais demande six comparaisons : la ruse arithmétique reste plus lisible.

```python
def trier_trois(a, b, c):
    """Retourne les trois valeurs dans l'ordre croissant."""
    mini = min(a, b, c)
    maxi = max(a, b, c)
    milieu = a + b + c - mini - maxi
    return (mini, milieu, maxi)

print(trier_trois(3, 1, 2))
```

```python
assert trier_trois(3, 1, 2) == (1, 2, 3)
assert trier_trois(1, 2, 3) == (1, 2, 3)
assert trier_trois(5, 5, 2) == (2, 5, 5)
assert trier_trois(-1, -7, 4) == (-7, -1, 4)
print("Solution 3 validée par les tests.")
```

### Solution de l'exercice 4 · Calculateur de mention

L'ordre des tests est décisif : on commence par le seuil le plus haut. Comme chaque `return` quitte la fonction, le premier seuil atteint donne la réponse et aucun test suivant ne s'exécute.

```python
def mention(note):
    """Retourne la mention associée à une note sur 20."""
    if note >= 16:
        return "Très bien"
    elif note >= 14:
        return "Bien"
    elif note >= 12:
        return "Assez bien"
    elif note >= 10:
        return "Passable"
    else:
        return "Insuffisant"

print(mention(13))
```

```python
assert mention(17) == "Très bien"
assert mention(16) == "Très bien"
assert mention(14) == "Bien"
assert mention(12) == "Assez bien"
assert mention(10) == "Passable"
assert mention(9.5) == "Insuffisant"
print("Solution 4 validée par les tests.")
```

### Solution de l'exercice 5 · Année bissextile

Trois tests suffisent quand on les ordonne du cas le plus particulier au plus général : 400, puis 100, puis 4. C'est exactement la logique SINON SI vue en cours, appliquée à une règle réelle.

```python
def est_bissextile(annee):
    """Applique les règles grégoriennes de l'année bissextile."""
    if annee % 400 == 0:
        return True
    if annee % 100 == 0:
        return False
    return annee % 4 == 0

print(est_bissextile(2000), est_bissextile(1900))
```

```python
assert est_bissextile(2024) is True
assert est_bissextile(2023) is False
assert est_bissextile(1900) is False
assert est_bissextile(2000) is True
assert est_bissextile(1996) is True
print("Solution 5 validée par les tests.")
```
