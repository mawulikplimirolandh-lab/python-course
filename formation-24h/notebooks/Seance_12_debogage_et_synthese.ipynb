### Exercice 3 · Intermédiaire · Chasse au bug : le décalage de un

La fonction `troncons(texte, taille)` doit découper un texte en morceaux de `taille` caractères. Sur `troncons("ABCDEF", 2)` elle retourne ['AB', 'CD', 'EF', ''] : un tronçon vide en trop. Exécutez, observez, corrigez la borne dans la cellule même.

```python
# Cette fonction contient un bug de borne : corrigez la
def troncons(texte, taille):
    resultat = []
    for debut in range(0, len(texte) + 1, taille):
        resultat.append(texte[debut:debut + taille])
    return resultat

print(troncons("ABCDEF", 2))   # attendu : ['AB', 'CD', 'EF']
```

*Vérification automatique (exécutez après avoir complété) :*

```python
# Vérification de l'exercice 3 (ne pas modifier)
try:
    assert troncons("ABCDEF", 2) == ["AB", "CD", "EF"], "cas pair"
    assert troncons("ABCDE", 2) == ["AB", "CD", "E"], "reste incomplet"
    assert troncons("A", 3) == ["A"], "texte plus court que la taille"
    assert troncons("", 2) == [], "texte vide"
    print("Exercice 3 : OK, les tests passent.")
except NameError:
    print("Exercice 3 : à compléter (fonction pas encore définie).")
except AssertionError as e:
    print("Exercice 3 : résultat incorrect pour le moment :", e)
except Exception as e:
    print("Exercice 3 : erreur pendant l'exécution :", type(e).__name__, e)
```

### Exercice 5 · Défi · Le programme mystère : cinq bugs

Le programme ci dessous calcule un bilan de ventes : il contient **cinq bugs** (deux d'exécution, trois logiques). Objectif : un rapport exact qui affiche, pour chaque produit, le chiffre d'affaires, puis le produit champion et le total général. Exécutez d'abord la cellule : l'enveloppe try/except du bas vous montre le premier incident. Corrigez ensuite un bug à la fois, dans la cellule même, en réexécutant après chaque correction. Les tests de vérification décrivent le résultat exact attendu : riz 4800 F, huile 5500 F, sucre 0 F, champion huile, total 10300 F.

```python
VENTES = {
    "riz": [(1200, 3), (1200, 1)],     # (prix unitaire, quantité)
    "huile": [(2000, 2), (1500, 1)],
    "sucre": [(1000, 0)],
}

def chiffre_affaires(ventes_produit):
    total = 0
    for prix, quantite in ventes_produit:
        total = total + prix - quantite          # bug 1
    return total

def rapport_ventes(ventes):
    lignes = []
    champion = None
    champion_ca = 0
    for produit, ventes_produit in ventes.items():
        ca = chiffre_affaires(ventes_produit)
        lignes.append(f"{produit} : {ca} F")
        if champion != None or ca > champion_ca:  # bug 2
            champion = produit
            champion_ca == ca                     # bug 3
    lignes.append(f"Champion : {champion} ({ca} F)")   # bug 4
    total_general = sum(ventes.values())          # bug 5
    lignes.append(f"Total général : {total_general} F")
    return "\n".join(lignes)

try:
    print(rapport_ventes(VENTES))
except Exception as incident:
    print("Le programme plante :", type(incident).__name__, "-", incident)
```

*Vérification automatique (exécutez après avoir complété) :*

```python
# Vérification de l'exercice 5 (ne pas modifier)
try:
    texte = rapport_ventes(VENTES)
    lignes = texte.split("\n")
    assert "riz : 4800 F" in lignes, "CA du riz (1200 x 3 + 1200 x 1)"
    assert "huile : 5500 F" in lignes, "CA de l'huile"
    assert "sucre : 0 F" in lignes, "CA du sucre"
    assert "Champion : huile (5500 F)" in lignes, "le champion est l'huile"
    assert "Total général : 10300 F" in lignes, "total de tous les CA"
    assert len(lignes) == 5, "trois produits + champion + total"
    print("Exercice 5 : OK, les cinq bugs sont corrigés.")
except NameError:
    print("Exercice 5 : à compléter (fonction pas encore corrigée).")
except AssertionError as e:
    print("Exercice 5 : il reste un bug :", e)
except Exception as e:
    print("Exercice 5 : erreur pendant l'exécution :", type(e).__name__, e)
```

### Solution de l'exercice 1 · Chasse au bug : la moyenne qui ment

Le coupable : `//` divise en entiers et tronque 36 // 3 en 12 (ok ici) mais 19 // 2 en 9 au lieu de 9.5. La division classique `/` est la bonne pour une moyenne, et elle retourne un float comme le veulent les tests. Symptôme, cause, correction : la méthode en quatre temps en deux minutes.

```python
def moyenne_boggee(valeurs):
    """Version corrigée : division classique."""
    total = 0
    for v in valeurs:
        total += v
    return total / len(valeurs)

print(moyenne_boggee([12, 15, 9]), moyenne_boggee([10, 9]))
```

```python
assert moyenne_boggee([12, 15, 9]) == 12.0
assert moyenne_boggee([10, 9]) == 9.5
assert isinstance(moyenne_boggee([10, 9]), float)
print("Solution 1 validée par les tests.")
```

### Solution de l'exercice 3 · Chasse au bug : le décalage de un

`range(0, len(texte) + 1, taille)` démarre un tour SUPPLÉMENTAIRE exactement à la fin du texte : le slice vide en témoigne. La borne correcte est `len(texte)` tout court : le piège du tableau des six pièges classiques, en direct.

```python
def troncons(texte, taille):
    """Version corrigée : la borne de fin est exclue."""
    resultat = []
    for debut in range(0, len(texte), taille):
        resultat.append(texte[debut:debut + taille])
    return resultat

print(troncons("ABCDEF", 2), troncons("ABCDE", 2))
```

```python
assert troncons("ABCDEF", 2) == ["AB", "CD", "EF"]
assert troncons("ABCDE", 2) == ["AB", "CD", "E"]
assert troncons("A", 3) == ["A"]
assert troncons("", 2) == []
print("Solution 3 validée par les tests.")
```

### Solution de l'exercice 4 · Quiz de synthèse auto corrigé

Le quiz illustre le dict en action (séance 06), le parcours avec .items(), la compréhension filtrante (séance 05) et la sécurité de .get contre les clés absentes. Notez le style : une fonction qui retourne un tuple (résultat, détail), le pattern des séances 05 et 06.

```python
# Le corrigé complet, avec la logique de correction
CORRIGE = {1: "b", 2: "b", 3: "c", 4: "b", 5: "b", 6: "a", 7: "b", 8: "c"}

def corriger_quiz(reponses):
    """Retourne (score, numéros manqués)."""
    score = sum(1 for n, bonne in CORRIGE.items()
                if reponses.get(n) == bonne)
    manquées = [n for n, bonne in CORRIGE.items()
                if reponses.get(n) != bonne]
    return (score, manquées)

parfait = dict(CORRIGE)
score, manquées = corriger_quiz(parfait)
print(f"Vérification du corrigé lui même : {score} / 8, manquées : {manquées}")
```

```python
CORRIGE = {1: "b", 2: "b", 3: "c", 4: "b",
           5: "b", 6: "a", 7: "b", 8: "c"}
assert corriger_quiz(dict(CORRIGE)) == (8, [])
assert corriger_quiz({}) == (0, [1, 2, 3, 4, 5, 6, 7, 8])
assert corriger_quiz({1: "b", 3: "b"}) == (1, [2, 3, 4, 5, 6, 7, 8])
print("Solution 4 validée par les tests.")
```

### Solution de l'exercice 5 · Le programme mystère : cinq bugs

Les cinq bugs : (1) `prix - quantite` au lieu de `prix * quantite` : tous les chiffres d'affaires sont faux (logique) ; (2) `champion != None` fait entrer dans la branche à CHAQUE tour : le champion devient le dernier produit parcouru (logique) ; (3) `champion_ca == ca` compare au lieu d'affecter : la ligne du champion lève NameError car champion_ca n'a jamais été rangé (exécution) ; (4) la ligne du champion affiche `ca` (celui du dernier produit) au lieu de `champion_ca` (logique) ; (5) `sum(ventes.values())` additionne des listes : TypeError (exécution). Le bon total parcourt les chiffres d'affaires : `sum(chiffre_affaires(v) for v in ventes.values())`. Trois logiques, deux d'exécution : c'était le contrat.

```python
def chiffre_affaires(ventes_produit):
    total = 0
    for prix, quantite in ventes_produit:
        total = total + prix * quantite      # bug 1 corrigé
    return total

def rapport_ventes(ventes):
    lignes = []
    champion = None
    champion_ca = 0
    for produit, ventes_produit in ventes.items():
        ca = chiffre_affaires(ventes_produit)
        lignes.append(f"{produit} : {ca} F")
        if champion is None or ca > champion_ca:   # bug 2 corrigé
            champion = produit
            champion_ca = ca                        # bug 3 corrigé
    lignes.append(f"Champion : {champion} ({champion_ca} F)")   # bug 4 corrigé
    total_general = sum(chiffre_affaires(v) for v in ventes.values())
    lignes.append(f"Total général : {total_general} F")          # bug 5 corrigé
    return "\n".join(lignes)

print(rapport_ventes(VENTES))
```
