### Solution de l'exercice 1 · Panier de courses

Après l'append et le remove, la liste vaut [riz, sucre, huile] ; le remplacement d'index 0 donne [pain, sucre, huile] (riz est remplacé, pas déplacé) ; le tri final ordonne alphabétiquement. `.sort()` trie sur place et retourne None : ne l'affectez jamais.

```python
def preparer_panier():
    """Applique les quatre opérations demandées sur le panier."""
    panier = ["savon", "riz", "sucre"]
    panier.append("huile")
    panier.remove("savon")
    panier[0] = "pain"
    panier.sort()
    return panier

print(preparer_panier())
```

```python
assert preparer_panier() == ["huile", "pain", "sucre"]
print("Solution 1 validée par les tests.")
```

### Solution de l'exercice 2 · Carrés et cubes en compréhension

Le gabarit est toujours le même : expression, for, éventuellement condition. `range(1, n + 1)` couvre les entiers de 1 à n inclus puisque la borne de fin est exclue.

```python
def carres(n):
    """Les carrés de 1 à n, en une ligne."""
    return [k ** 2 for k in range(1, n + 1)]

def cubes(n):
    """Les cubes de 1 à n, en une ligne."""
    return [k ** 3 for k in range(1, n + 1)]

print(carres(5))
print(cubes(4))
```

```python
assert carres(5) == [1, 4, 9, 16, 25]
assert cubes(4) == [1, 8, 27, 64]
assert carres(0) == []
print("Solution 2 validée par les tests.")
```

### Solution de l'exercice 3 · Filtrer les notes admises

La compréhension fait le filtrage en une ligne ; le SI protège la moyenne du cas aucun admis, qui provoquerait une division par zéro. 41 divisé par 3 donne 13.666... arrondi à 13.67.

```python
def bilan(notes):
    """Retourne (admis, moyenne_admis) arrondie à 2 décimales."""
    admis = [n for n in notes if n >= 10]
    if admis:
        moyenne_admis = round(sum(admis) / len(admis), 2)
    else:
        moyenne_admis = None
    return (admis, moyenne_admis)

print(bilan([12, 8, 15, 9, 14]))
```

```python
assert bilan([12, 8, 15, 9, 14]) == ([12, 15, 14], 13.67)
assert bilan([8, 9]) == ([], None)
assert bilan([10]) == ([10], 10.0)
print("Solution 3 validée par les tests.")
```

### Solution de l'exercice 4 · Transposée d'une matrice

Le for externe balaie les colonnes (j de 0 à largeur moins 1) ; le for interne, écrit en premier dans l'expression, collecte la valeur de la colonne j dans chaque ligne. Vérifiez l'ordre mental : pour chaque j, pour chaque ligne, prendre ligne[j].

```python
def transposee(m):
    """Transposée par compréhension imbriquée."""
    return [[ligne[j] for ligne in m] for j in range(len(m[0]))]

print(transposee([[1, 2, 3], [4, 5, 6]]))
```

```python
assert transposee([[1, 2, 3], [4, 5, 6]]) == [[1, 4], [2, 5], [3, 6]]
assert transposee([[7]]) == [[7]]
print("Solution 4 validée par les tests.")
```

### Solution de l'exercice 5 · Nettoyage de données imbriquées

L'astuce tient dans le second for : chaque élément est transformé en liste à un élément s'il ne l'est pas déjà, ce qui permet un parcours uniforme. `isinstance` évite le piège du `type(x) == int` qui exclurait les classes filles. Attention : `True` passe le test car bool est une spécialisation de int, comportement assumé ici.

```python
def nettoyer(donnees):
    """Extrait tous les nombres d'une structure arbitrairement sale."""
    return [element
            for sous_liste in donnees
            for element in (sous_liste if isinstance(sous_liste, list)
                            else [sous_liste])
            if isinstance(element, (int, float))]

print(nettoyer([1, "a", [2, None], "b", [3, 4.5]]))
```

```python
assert nettoyer([1, "a", [2, None], "b", [3, 4.5]]) == [1, 2, 3, 4.5]
assert nettoyer([]) == []
assert nettoyer(["x", None]) == []
print("Solution 5 validée par les tests.")
```
