### Solution de l'exercice 1 · Vehicule et Voiture

La fille fixe elle même roues = 4 : l'appelant ne répète jamais une information déjà connue. La redéfinition de decrire() s'appuie sur super().decrire() : si le format du parent change un jour, la fille suit automatiquement.

```python
class Vehicule:
    """Véhicule générique."""

    def __init__(self, marque, roues):
        self.marque = marque
        self.roues = roues

    def decrire(self):
        return f"{self.marque} : {self.roues} roues"

class Voiture(Vehicule):
    """Une voiture EST UN véhicule à quatre roues."""

    def __init__(self, marque):
        super().__init__(marque, roues=4)

    def decrire(self):
        return super().decrire() + " (voiture)"

print(Vehicule("Generique", 3).decrire())
print(Voiture("Toyota").decrire())
```

```python
assert Voiture("Toyota").decrire() == "Toyota : 4 roues (voiture)"
assert Vehicule("G", 3).decrire() == "G : 3 roues"
assert isinstance(Voiture("X"), Vehicule)
print("Solution 1 validée par les tests.")
```

### Solution de l'exercice 2 · __str__ et __repr__ sur Media

print(m) appelle __str__ ; imprimer une liste appelle __repr__ de chaque élément : c'est pour cela que le __repr__ doit montrer comment recréer l'objet. Règle du cours : si vous n'écrivez qu'un seul des deux, écrivez __repr__, Python l'utilise en secours pour __str__.

```python
class Media:
    """Un média avec titre et durée en minutes."""

    def __init__(self, titre, duree):
        self.titre = titre
        self.duree = duree

    def __str__(self):
        return f"{self.titre} ({self.duree} min)"

    def __repr__(self):
        return f"Media('{self.titre}', {self.duree})"

m = Media("Sira", 104)
print(m)
print([m])
```

### Solution de l'exercice 3 · Formes polymorphes

`total_aire` ignore complètement les classes concrètes : elle ne connaît que le contrat `aire()`. Ajouter demain un Triangle ne modifiera aucune ligne de total_aire : c'est la définition même du polymorphisme, et la porte d'entrée vers les classes abstraites du Tome 2.

```python
class Forme:
    """Contrat de base : toute forme sait calculer son aire."""

    def aire(self):
        raise NotImplementedError("implémentez aire()")

class Cercle(Forme):
    """Un cercle EST UNE forme."""

    def __init__(self, rayon):
        self.rayon = rayon

    def aire(self):
        return 3.14159 * self.rayon ** 2

class Rectangle(Forme):
    """Un rectangle EST UNE forme."""

    def __init__(self, largeur, hauteur):
        self.largeur = largeur
        self.hauteur = hauteur

    def aire(self):
        return self.largeur * self.hauteur

def total_aire(formes):
    """Somme polymorphe : chaque forme répond avec sa formule."""
    total = 0
    for forme in formes:
        total += forme.aire()
    return total

print(total_aire([Cercle(1), Rectangle(2, 3)]))
```

```python
formes = [Cercle(1), Rectangle(2, 3), Cercle(2)]
assert abs(total_aire(formes) - 21.70795) < 1e-4
assert abs(Cercle(1).aire() - 3.14159) < 1e-5
print("Solution 3 validée par les tests.")
```

### Solution de l'exercice 4 · Classe Duree avec __add__ et __eq__

Tout repose sur le stockage normalisé en minutes : comparaisons et additions deviennent triviales. __add__ retourne une **nouvelle** Duree plutôt que de modifier self : les opérateurs arithmétiques ne doivent jamais avoir d'effet de bord sur leurs opérandes.

```python
class Duree:
    """Une durée stockée en minutes, toujours normalisée."""

    def __init__(self, heures, minutes=0):
        self.total = heures * 60 + minutes

    def __add__(self, autre):
        total = self.total + autre.total
        return Duree(total // 60, total % 60)

    def __eq__(self, autre):
        return self.total == autre.total

    def __lt__(self, autre):
        return self.total < autre.total

    def __repr__(self):
        return f"Duree({self.total // 60}, {self.total % 60})"

print(Duree(1, 90) == Duree(2, 30))
print(Duree(1, 90) + Duree(0, 45))
```

```python
assert Duree(1, 90) == Duree(2, 30)
assert (Duree(1, 90) + Duree(0, 45)).total == 195
assert repr(Duree(1, 90)) == "Duree(2, 30)"
assert Duree(1, 0) < Duree(2, 0)
print("Solution 4 validée par les tests.")
```

### Solution de l'exercice 5 · Bibliothèque de médias indexable

Les filles spécialisent __repr__ sans toucher au reste : l'héritage économise tout le commun. La Bibliotheque délègue __len__ et __getitem__ à sa liste interne : trois lignes suffisent pour transformer une classe en conteneur natif, iterable par le for de l'exercice.

```python
class Media:
    """Classe de base des médias."""

    def __init__(self, titre, duree):
        self.titre = titre
        self.duree = duree

    def __repr__(self):
        return f"Media('{self.titre}', {self.duree})"

class Livre(Media):
    """Un livre EST UN média avec un auteur."""

    def __init__(self, titre, auteur):
        super().__init__(titre, duree=0)
        self.auteur = auteur

    def __repr__(self):
        return f"Livre('{self.titre}', '{self.auteur}')"

class Film(Media):
    """Un film EST UN média d'une durée donnée."""

    def __repr__(self):
        return f"Film('{self.titre}', {self.duree})"

class Bibliotheque:
    """Conteneur de médias : len(), index et boucle for."""

    def __init__(self):
        self._medias = []

    def ajouter(self, media):
        self._medias.append(media)

    def __len__(self):
        return len(self._medias)

    def __getitem__(self, index):
        return self._medias[index]

bib = Bibliotheque()
bib.ajouter(Livre("L'Enfant noir", "Laye"))
bib.ajouter(Film("Sira", 104))
print(len(bib), bib[0])
for media in bib:
    print(media)
```

```python
bib = Bibliotheque()
bib.ajouter(Livre("L'Enfant noir", "Laye"))
bib.ajouter(Film("Sira", 104))
assert len(bib) == 2
assert repr(bib[0]) == "Livre('L'Enfant noir', 'Laye')"
assert repr(bib[1]) == "Film('Sira', 104)"
assert [media.titre for media in bib] == ["L'Enfant noir", "Sira"]
print("Solution 5 validée par les tests.")
```
