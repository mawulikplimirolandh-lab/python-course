### Solution de l'exercice 1 · Étape 1 : la classe Client

Le compteur vit sur la classe (partagé), les identités vivent sur les instances. C'est la grille de lecture de la séance 08 appliquée au projet : attribut de classe pour ce qui est global, d'instance pour ce qui est individuel.

```python
class Client:
    """Fiche d'identité d'un client de la banque."""

    total_clients = 0

    def __init__(self, numero, nom):
        self.numero = numero
        self.nom = nom
        Client.total_clients += 1

    @classmethod
    def combien(cls):
        return cls.total_clients

    def __str__(self):
        return f"Client {self.numero} : {self.nom}"

print(Client(1, "Awa"))
```

### Solution de l'exercice 2 · Étape 2 : la classe Compte avec historique

Toute modification passe par les deux méthodes validées : la propriété en lecture seule verrouille le solde, l'historique raconte chaque opération avec son signe. Le __str__ lit client.nom : le Compte connaît son Client par composition (A UN client).

```python
class Compte:
    """Compte validé avec fil d'audit complet."""

    def __init__(self, numero, client, solde=0):
        self.numero = numero
        self.client = client
        self._solde = solde
        self.historique = [f"Ouverture : {solde} F"]

    @property
    def solde(self):
        return self._solde

    def deposer(self, montant):
        if montant <= 0:
            raise ValueError("le dépôt doit être strictement positif")
        self._solde += montant
        self.historique.append(f"Dépôt : +{montant} F")

    def retirer(self, montant):
        if montant <= 0:
            raise ValueError("le retrait doit être strictement positif")
        if montant > self._solde:
            raise ValueError("solde insuffisant")
        self._solde -= montant
        self.historique.append(f"Retrait : -{montant} F")

    def __str__(self):
        return f"Compte {self.numero} ({self.client.nom}) : {self._solde} F"

awa = Client(1, "Awa")
c = Compte(10, awa, 1000)
c.deposer(500)
print(c)
for ligne in c.historique:
    print(" ", ligne)
```

```python
awa = Client(1, "Awa")
c = Compte(11, awa, 500)
c.deposer(250); c.retirer(100)
assert c.solde == 650
assert len(c.historique) == 3
try:
    c.retirer(9999); refuse = False
except ValueError:
    refuse = True
assert refuse
print("Solution 2 validée par les tests.")
```

### Solution de l'exercice 3 · Étape 3 : les virements

Le virement ne touche jamais `_solde` : il réutilise les méthodes validées, donc l'ordre des traces reste correct et un solde insuffisant bloque tout avant tout crédit. `source is cible` compare les identités d'objets, pas leurs valeurs : exactement le rôle de l'opérateur is vu en séance 05.

```python
def virer(source, cible, montant):
    """Virement validé : retire de la source, crédite la cible."""
    if montant <= 0:
        raise ValueError("montant invalide")
    if source is cible:
        raise ValueError("source et cible identiques")
    source.retirer(montant)
    cible.deposer(montant)
    return source.solde

awa = Client(1, "Awa")
kofi = Client(2, "Kofi")
c1 = Compte(20, awa, 1000)
c2 = Compte(21, kofi, 500)
print(virer(c1, c2, 400), c1.solde, c2.solde)
```

```python
awa = Client(1, "Awa")
kofi = Client(2, "Kofi")
c1 = Compte(22, awa, 1000)
c2 = Compte(23, kofi, 500)
assert virer(c1, c2, 400) == 600
assert (c1.solde, c2.solde) == (600, 900)
try:
    virer(c1, c1, 10); refuse = False
except ValueError:
    refuse = True
assert refuse
print("Solution 3 validée par les tests.")
```

### Solution de l'exercice 4 · Étape 4 : la classe Banque

La Banque fabrique elle même les numéros : l'appelant ne peut plus créer de collision. `KeyError` avec un message clair vaut mieux qu'un silence : la séance 11 la remplacera par une CompteInconnuError dédiée. Le rapport réutilise str(compte) : le format n'existe qu'à un endroit.

```python
class Banque:
    """Orchestrateur : ouverture, recherche, totaux, rapport."""

    def __init__(self):
        self.comptes = {}
        self.prochain_numero = 1

    def ouvrir_compte(self, nom, solde=0):
        numero = self.prochain_numero
        client = Client(numero, nom)
        compte = Compte(numero, client, solde)
        self.comptes[numero] = compte
        self.prochain_numero += 1
        return compte

    def compte(self, numero):
        if numero not in self.comptes:
            raise KeyError(f"aucun compte numéro {numero}")
        return self.comptes[numero]

    def total_depots(self):
        return sum(c.solde for c in self.comptes.values())

    def rapport(self):
        lignes = ["Rapport de la banque"]
        for compte in self.comptes.values():
            lignes.append(str(compte))
        lignes.append(f"Total des dépôts : {self.total_depots()} F")
        return "\n".join(lignes)

b = Banque()
b.ouvrir_compte("Awa", 1000)
b.ouvrir_compte("Kofi", 250)
print(b.rapport())
```

```python
b = Banque()
c1 = b.ouvrir_compte("X", 100)
c2 = b.ouvrir_compte("Y", 200)
assert b.total_depots() == 300
assert b.compte(2) is c2
try:
    b.compte(77); absent = False
except KeyError:
    absent = True
assert absent
print("Solution 4 validée par les tests.")
```

### Solution de l'exercice 5 · Étape 5 : intérêts et rapport mensuel

`round(solde * taux)` arrondit à l'unité de franc : la règle métier est écrite une seule fois, visible et testable. Le compte vide est explicitement épargné par le SI. Dans votre version finale, recopiez la méthode à l'intérieur de la classe Compte ; la greffe `Compte.appliquer_interets = ...` ci dessous est l'astuce qui rend la cellule autonome sans dupliquer les classes.

```python
# La méthode doit vivre DANS la classe Compte : on la greffe ici
def appliquer_interets(self, taux):
    """Crédite les intérêts d'un compte créditeur, avec trace."""
    if self._solde > 0:
        gain = round(self._solde * taux)
        self._solde += gain
        self.historique.append(f"Intérêts : +{gain} F")

Compte.appliquer_interets = appliquer_interets

def cloture_mensuelle(banque, taux):
    """Crédite les intérêts de chaque compte, puis édite le rapport."""
    for compte in banque.comptes.values():
        compte.appliquer_interets(taux)
    return banque.rapport()

b = Banque()
b.ouvrir_compte("Awa", 1000)
b.ouvrir_compte("Kofi", 0)
print(cloture_mensuelle(b, 0.05))
```

```python
b = Banque()
c1 = b.ouvrir_compte("A", 1000)
c2 = b.ouvrir_compte("B", 0)
cloture_mensuelle(b, 0.05)
assert c1.solde == 1050 and c2.solde == 0
assert c1.historique[-1] == "Intérêts : +50 F"
print("Solution 5 validée par les tests.")
```
