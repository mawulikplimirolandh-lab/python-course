### Solution de l'exercice 1 · Classe Point

`self.x = x` : à gauche l'attribut de l'instance, à droite le paramètre reçu. La distance utilise l'exposant 0.5 au lieu de math.sqrt : les deux sont équivalents ici. __str__ retourne toujours une chaîne.

```python
class Point:
    """Un point du plan avec sa distance à l'origine."""

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self):
        return (self.x ** 2 + self.y ** 2) ** 0.5

    def __str__(self):
        return f"Point({self.x}, {self.y})"

p = Point(3, 4)
print(p, p.distance())
```

```python
p = Point(3, 4)
assert p.distance() == 5.0
assert str(p) == "Point(3, 4)"
assert Point(0, 0).distance() == 0.0
print("Solution 1 validée par les tests.")
```

### Solution de l'exercice 3 · CompteSimple validé

Le souligné simple `_solde` est une convention de respect, pas une interdiction technique : le vrai verrou est l'absence de setter. Toute modification passe par deposer ou retirer, qui valident : impossible de créer un solde incohérent. À la séance 11, ces ValueError deviendront des exceptions métier dédiées.

```python
class CompteSimple:
    """Compte dont le solde ne change que par des méthodes validées."""

    def __init__(self, titulaire, solde=0):
        self.titulaire = titulaire
        self._solde = solde

    @property
    def solde(self):
        return self._solde

    def deposer(self, montant):
        if montant <= 0:
            raise ValueError("le dépôt doit être strictement positif")
        self._solde += montant

    def retirer(self, montant):
        if montant <= 0:
            raise ValueError("le retrait doit être strictement positif")
        if montant > self._solde:
            raise ValueError("solde insuffisant")
        self._solde -= montant

    def __str__(self):
        return f"Compte de {self.titulaire} : {self._solde} F"

c = CompteSimple("Awa", 1000)
c.deposer(500)
print(c)
```

```python
c = CompteSimple("Awa", 1000)
c.deposer(500); c.retirer(300)
assert c.solde == 1200
try:
    c.retirer(-5); refuse = False
except ValueError:
    refuse = True
assert refuse
print("Solution 3 validée par les tests.")
```

### Solution de l'exercice 4 · Température avec @property

Même __init__ passe par le setter : la règle du zéro absolu est vérifiée dès la création, pas seulement aux modifications. `fahrenheit` n'a pas de setter : propriété calculée en lecture seule, dérivée de celsius.

```python
class Temperature:
    """Température en Celsius, jamais sous le zéro absolu."""

    def __init__(self, celsius=0):
        self.celsius = celsius   # déclenche le setter

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, valeur):
        if valeur < -273.15:
            raise ValueError("sous le zéro absolu : impossible")
        self._celsius = valeur

    @property
    def fahrenheit(self):
        return self._celsius * 9 / 5 + 32

t = Temperature(25)
print(t.fahrenheit)
```

```python
t = Temperature(25)
assert t.fahrenheit == 77.0
t.celsius = -40
assert t.fahrenheit == -40.0
try:
    t.celsius = -300; refuse = False
except ValueError:
    refuse = True
assert refuse
print("Solution 4 validée par les tests.")
```

### Solution de l'exercice 5 · Playlist avec compteur de créations

L'incrément `Playlist.nombre_plists += 1` touche l'attribut de classe, partagé par toutes les instances ; `_titres` naît dans __init__, propre à chaque objet : le contraste vu en cours appliqué aux deux extrémités. `cls` dans combien pointe vers Playlist.

```python
class Playlist:
    """Playlist de titres avec compteur global de créations."""

    nombre_plists = 0

    def __init__(self, nom):
        self.nom = nom
        self._titres = []
        Playlist.nombre_plists += 1

    @classmethod
    def combien(cls):
        return cls.nombre_plists

    def ajouter(self, titre):
        self._titres.append(titre)

    @property
    def duree_estimee(self):
        return 3 * len(self._titres)

    def __str__(self):
        return f"Playlist {self.nom} : {len(self._titres)} titres"

p1 = Playlist("Afrobeat")
p1.ajouter("Jerusalema")
print(p1, "| playlists créées :", Playlist.combien())
```

```python
av = Playlist.combien()
a, b = Playlist("X"), Playlist("Y")
assert Playlist.combien() == av + 2
a.ajouter("Titre1"); a.ajouter("Titre2"); b.ajouter("Titre3")
assert a.duree_estimee == 6 and b.duree_estimee == 3
assert b._titres == ["Titre3"]
print("Solution 5 validée par les tests.")
```
