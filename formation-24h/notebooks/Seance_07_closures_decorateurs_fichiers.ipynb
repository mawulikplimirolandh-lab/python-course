### Solution de l'exercice 1 · Nettoyage de chaîne

Le trésor caché : `split()` **sans argument** découpe sur tout bloc d'espaces, quel que soit son épaisseur, et ne produit aucun élément vide. Le join recolle avec exactement un espace. Trois méthodes enchaînées, une ligne.

```python
def normaliser(phrase):
    """Minuscules, bords propres, un seul espace entre les mots."""
    return " ".join(phrase.strip().lower().split())

print(normaliser("  BONJOUR   la  CLASSE  "))
```

```python
assert normaliser("  BONJOUR   la  CLASSE  ") == "bonjour la classe"
assert normaliser("RIZ") == "riz"
print("Solution 1 validée par les tests.")
```

### Solution de l'exercice 2 · Palindrome

La compréhension garde chaque caractère alphanumérique en minuscule ; le join les recolle. La tranche `[::-1]` renverse la chaîne. Attention aux accents : « é » n'est pas « e » pour isalnum, le fameux vers d'Esope ne passe qu'en version sans accent.

```python
def est_palindrome(phrase):
    """True si la phrase se lit dans les deux sens, ponctuation ignorée."""
    nette = "".join(c.lower() for c in phrase if c.isalnum())
    return nette == nette[::-1]

print(est_palindrome("Esope reste ici et se repose"))
```

```python
assert est_palindrome("kayak") is True
assert est_palindrome("Esope reste ici et se repose") is True
assert est_palindrome("Bonjour") is False
print("Solution 2 validée par les tests.")
```

### Solution de l'exercice 3 · Fabrique de puissances (closure)

Chaque appel de `fabrique_puissance` crée une nouvelle cellule qui garde son exposant : c'est la preuve vivante que la closure retient la variable, pas seulement sa valeur au moment du return.

```python
def fabrique_puissance(exposant):
    """Fabrique une closure qui élève à la puissance donnée."""
    def appliquer(x):
        return x ** exposant
    return appliquer

carre = fabrique_puissance(2)
cube = fabrique_puissance(3)
print(carre(5), cube(5))
print("Exposant retenu par carre :",
      carre.__closure__[0].cell_contents)
```

```python
assert carre(5) == 25
assert cube(5) == 125
assert carre.__closure__[0].cell_contents == 2
assert cube.__closure__[0].cell_contents == 3
print("Solution 3 validée par les tests.")
```

### Solution de l'exercice 4 · Décorateur compteur d'appels

L'attribut `emballage.appels` joue le rôle d'un compteur rangé dans la closure, accessible de l'extérieur. @wraps est vital : sans lui, le nom serait « emballage » et les outils de documentation se perdraient.

```python
from functools import wraps

def compte_appels(fonction):
    """Décorateur qui compte les appels dans .appels."""
    @wraps(fonction)
    def emballage(*args, **kwargs):
        emballage.appels += 1
        return fonction(*args, **kwargs)
    emballage.appels = 0
    return emballage

@compte_appels
def saluer(nom):
    """Retourne une salutation."""
    return f"Bonjour {nom} !"

print(saluer.appels, saluer.__name__)   # 0 : compteur au depart
```

```python
saluer("X"); saluer("Y"); saluer("Z")
assert saluer.appels == 3
assert saluer.__name__ == "saluer"
print("Solution 4 validée par les tests.")
```

### Solution de l'exercice 5 · Journal de bord dans un fichier

Trois temps séparés : écrire, relire, compter. Chaque bloc with garantit la fermeture. Le comptage réutilise le schéma `get(cle, 0) + 1` de la séance 06. Le fichier disparaît à la fin : une fonction qui nettoie après elle est rejouable sans surprise.

```python
import os

def tenir_journal(evenements):
    """Écrit un journal, le relit et compte chaque événement."""
    with open("journal.txt", "w") as f:
        for evenement in evenements:
            f.write(evenement + "\n")

    compte = {}
    with open("journal.txt", "r") as f:
        for ligne in f:
            evenement = ligne.strip()
            compte[evenement] = compte.get(evenement, 0) + 1

    os.remove("journal.txt")
    return compte

print(tenir_journal(["ouverture", "vente", "ouverture",
                     "clôture", "vente", "ouverture"]))
```

```python
essai = ["ouverture", "vente", "ouverture",
         "clôture", "vente", "ouverture"]
assert tenir_journal(essai) == {"ouverture": 3, "vente": 2, "clôture": 1}
import os
assert not os.path.exists("journal.txt")
print("Solution 5 validée par les tests.")
```
