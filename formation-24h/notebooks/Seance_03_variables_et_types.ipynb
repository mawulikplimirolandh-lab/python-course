### Solution de l'exercice 2 · Convertisseur secondes vers h:m:s

On découpe du plus gros au plus fin : les heures d'abord, puis on garde le reste pour les minutes, puis les secondes. C'est le pattern général du changement d'unité.

```python
def convertir(secondes):
    """Décompose un total de secondes en (heures, minutes, secondes)."""
    h = secondes // 3600
    reste = secondes % 3600
    m = reste // 60
    s = reste % 60
    return (h, m, s)

print(convertir(7385))
```

```python
assert convertir(7385) == (2, 3, 5)
assert convertir(3600) == (1, 0, 0)
assert convertir(59) == (0, 0, 59)
assert convertir(0) == (0, 0, 0)
print("Solution 2 validée par les tests.")
```

### Solution de l'exercice 3 · Prix TTC avec arrondi

Multiplier par `(1 + taux)` évite d'écrire deux fois le prix. L'arrondi `round(x, 2)` arrange l'affichage monétaire, mais retenez que les floats restent des approximations binaires : les systèmes bancaires utilisent d'autres techniques, vues au Tome 2.

```python
def prix_ttc(hors_taxe, taux=0.18):
    """Retourne le prix TTC arrondi à deux décimales."""
    return round(hors_taxe * (1 + taux), 2)

for ht in (100, 1999.99):
    print(f"{ht:>10.2f} F hors taxe -> {prix_ttc(ht):>10.2f} F TTC")
```

```python
assert prix_ttc(100) == 118.0
assert prix_ttc(100, 0.075) == 107.5
assert prix_ttc(1999.99) == 2359.99
assert prix_ttc(0) == 0.0
print("Solution 3 validée par les tests.")
```

### Solution de l'exercice 4 · Échange de variables et alignement

L'affectation multiple `a, b = b, a` échange sans variable temporaire : Python évalue le membre droit en premier puis range. Les formats `:<8` et `:>10.2f` fixent la largeur et l'alignement : idéal pour des états lisibles en console.

```python
def echange(a, b):
    """Échange les deux valeurs à la façon Python."""
    a, b = b, a
    return a, b

def tableau(prenom, solde):
    """Une ligne de tableau alignée avec les formats de f-string."""
    return f"| {prenom:<8} | {solde:>10.2f} |"

print(echange(1, 2))
print(tableau("Awa", 1500))
print(tableau("Kossi", 98765.5))
```

```python
assert echange(1, 2) == (2, 1)
assert echange("x", 3) == (3, "x")
assert tableau("Awa", 1500) == "| Awa      |    1500.00 |"
print("Solution 4 validée par les tests.")
```

### Solution de l'exercice 5 · Rendre la monnaie en FCFA

L'algorithme glouton prend d'abord la plus grande coupure possible, autant de fois que possible, puis descend d'un cran. Pour le système monétaire du franc CFA, il donne toujours le minimum de coupures. Le dictionnaire résultat ne contient que les coupures utilisées.

```python
COUPURES = [10000, 5000, 2000, 1000, 500, 100, 50, 25, 10, 5, 1]

def rendre_montant(total, paye):
    """Rend la monnaie avec le moins de coupures (méthode gloutonne)."""
    if paye < total:
        raise ValueError("paiement insuffisant")
    reste = paye - total
    monnaie = {}
    for coupure in COUPURES:
        nombre = reste // coupure
        if nombre > 0:
            monnaie[coupure] = nombre
            reste = reste % coupure
    return monnaie

print(rendre_montant(0, 2350))
```

```python
assert rendre_montant(0, 2350) == {2000: 1, 100: 3, 50: 1}
assert rendre_montant(500, 500) == {}
assert rendre_montant(0, 75) == {50: 1, 25: 1}
assert rendre_montant(0, 10000) == {10000: 1}
print("Solution 5 validée par les tests.")
```
